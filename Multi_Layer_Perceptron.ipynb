## 12. Multi-Layered Neural Networks & Forward Propagation

image.png

### **Overview of a Multi-Layered Neural Network (2-Layered NN)**
Unlike single-layer perceptrons, a **Multi-Layered Neural Network** consists of:
1. **Input Layer:** Receives the raw feature values.
2. **Hidden Layer (HL1):** Contains one or more hidden neurons that extract intermediate representations and patterns.
3. **Output Layer:** Combines hidden layer outputs to produce the final prediction.

---

### **Example Dataset: Predicting Exam Results**

Let's look at a dataset with three input features for students:
* **$x_1$ (IQ):** e.g., $95, 100, 95$
* **$x_2$ (Study Hours):** e.g., $4, 5, 2$
* **$x_3$ (Play Hours):** e.g., $4, 2, 7$
* **Output (Pass/Fail):** $1$ (Pass) or $0$ (Fail)

| IQ ($x_1$) | Study Hours ($x_2$) | Play Hours ($x_3$) | Output ($O/P$) |
| :---: | :---: | :---: | :---: |
| 95 | 4 | 4 | 1 |
| 100 | 5 | 2 | 1 |
| 95 | 2 | 7 | 0 |

---

### **Forward Propagation Process**
**Forward Propagation** is the step-by-step process of passing input data through the network from the input layer, through the hidden layers, to generate an output at the final layer.

For any given neuron in the hidden layer, forward propagation happens in two core steps:

#### **Step 1: Weighted Sum & Bias Calculation (Linear Transformation)**
Each input is multiplied by its corresponding weight, and a bias term is added:
$$Z = x_1 w_1 + x_2 w_2 + x_3 w_3 + b_1$$

Expressed in summation / vector form:
$$Z = \sum_{i=1}^{n} w^T x_i + b$$

#### **Step 2: Activation Function**
The linear output $Z$ is then passed through an **Activation Function** to introduce non-linearity:
$$\text{Activation}(Z)$$

## 13. Forward Propagation Calculation & Sigmoid Activation Function

### **Manual Numerical Example (Hidden Layer 1)**

Using our first student row from the dataset:
* **Inputs:** $x_1 = 95$ (IQ), $x_2 = 4$ (Study Hours), $x_3 = 4$ (Play Hours)
* **Weights:** $w_1 = 0.01, \, w_2 = 0.02, \, w_3 = 0.03$
* **Bias:** $b = 0.001$


### **Step 1: Weighted Sum Calculation ($Z$)**
$$Z = (95 \times 0.01) + (4 \times 0.02) + (4 \times 0.03) + (1 \times 0.001)$$
$$Z = 0.95 + 0.08 + 0.12 + 0.001 = 1.151$$


### **Step 2: Activation Function (Sigmoid)**
The linear output $Z$ is passed through the **Sigmoid Activation Function**, which squashes the value to a probability between $0$ and $1$.

#### **Sigmoid Formula:**
$$f(z) = \frac{1}{1 + e^{-z}}$$

#### **Calculation:**
$$f(1.151) = \frac{1}{1 + e^{-1.151}} \approx 0.759$$

* **Interpretation:** The neuron outputs a probability of **$0.759$** (or $75.9\%$), indicating a strong likelihood of passing!

## 14. Forward Propagation: Next Layer / Output Layer Calculation

In the previous step, the first neuron computed an intermediate output:
$$O_1 = 0.759$$

Now, this value $O_1$ is passed as the input to the next layer / output neuron with its own assigned weight ($w_4$) and bias ($b_2$).


### **Given Parameters**
* **Input from previous neuron ($O_1$):** $0.759$
* **Weight ($w_4$):** $0.02$
* **Bias ($b_2$):** $0.03$


### **Step 1: Weighted Sum Calculation ($Z$)**
$$Z = O_1 \times w_4 + b_2$$
$$Z = (0.759 \times 0.02) + 0.03$$
$$Z = 0.01518 + 0.03 = 0.04518$$



### **Step 2: Activation Function (Sigmoid)**
The linear result $Z = 0.04518$ is passed through the Sigmoid activation function to obtain the final output ($O_2$):

$$O_2 = \frac{1}{1 + e^{-Z}} = \frac{1}{1 + e^{-(0.04518)}}$$

$$\frac{1}{1 + 0.95582} \approx 0.5113$$

* **Final Output ($O_2$):** $\approx 0.5113$ (or $51.13\%$)


### **Summary of Forward Pass**
1. Raw features ($x_1, x_2, x_3$) entered the input layer.
2. Hidden Layer 1 processed the features to produce $O_1 = 0.759$.
3. Output Layer processed $O_1$ using weight $w_4$ and bias $b_2$ to generate the final network prediction $O_2 \approx 0.5113$.

## 15. Forward Propagation, Loss Function, and Backward Propagation

### **1. Forward Propagation**
* **Definition:** The process of passing inputs ($x_1, x_2, x_3$) forward through the neural network layers (from the input layer, through Hidden Layer 1, to the Output Layer) to generate a prediction.
* **Calculation Flow:**
  1. Computes the weighted sum and bias ($Z = \sum w_i x_i + b$).
  2. Applies the activation function to generate neuron outputs ($O_1, O_2$).


### **2. Loss Function (Error Calculation)**
Once forward propagation finishes and the network produces a prediction ($O_2 \approx 0.51129$), we compare it against the actual ground truth target value ($1$).

* **Formula / Calculation shown in lecture:**
  $$\text{Loss} = \text{Actual Target} - \text{Predicted Output}$$
  $$\text{Loss} = 1 - 0.51129 \approx 0.48871 \ (\approx 0.49)$$

* **Purpose:** The loss function measures how wrong or inaccurate the model's prediction is. A higher loss means the model needs significant adjustment.


### **3. Backward Propagation**
* **Definition:** Once the error (loss) is computed at the output layer, **Backward Propagation** travels backward through the network (from the output layer back to the input weights).
* **Purpose:** It uses calculus (gradients/derivatives) to determine how much each weight ($w_1, w_2, w_3, w_4$) and bias contributed to the error.
* **Optimization:** Using this error information, the network updates the weights to minimize the loss in future iterations, making predictions more accurate.
