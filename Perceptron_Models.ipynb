## 11. Perceptron Models: Single-Layer vs. Multi-Layer Perceptrons

As shown in the lecture diagram, Perceptron models are broadly categorized into two types:
1. **Single Layer Perceptron Model (SLP)**
2. **Multi-Layer Perceptron Model (MLP)** (which forms the core structure of an **Artificial Neural Network - ANN**)


### **1. Single-Layer Perceptron (SLP)**
* **Structure:** Consists of a single layer of output nodes where inputs are fed directly to the outputs through a set of weights. It uses a linear threshold/activation function.
* **Capability:** Can only learn linearly separable patterns (e.g., basic logic gates like AND, OR).
* **Limitation:** Fails entirely on non-linearly separable problems (such as the classic XOR problem).

### **2. Multi-Layer Perceptron (MLP / ANN)**
* **Structure:** Contains one or more **hidden layers** between the input layer and the output layer, along with non-linear activation functions.
* **Capability:** Can learn complex, non-linear relationships and patterns in data.


### **Advantages and Disadvantages of Perceptron Models**

#### **Advantages**
* **Simplicity:** Single-layer perceptrons are very easy to understand, implement, and train for basic linear tasks.
* **Foundation of Deep Learning:** They form the fundamental building block for complex modern neural networks (ANNs, CNNs, Transformers).
* **Real-Time Learning:** They can adjust their weights dynamically when processing new data sequentially (using the Perceptron Learning Rule).

#### **Disadvantages**
* **Inability to Solve Non-Linear Problems (SLP):** A single-layer perceptron cannot solve non-linear problems like the XOR function because it can only draw a single straight decision boundary.
* **Sensitivity to Noise:** Performance can degrade if the training data contains extreme outliers or noise unless properly regularized.
* **Limited Processing Power for Complex Tasks:** Single-layer models cannot handle complex data types like high-resolution images or raw natural language text without multi-layered deep architectures (MLPs).


### **Example: Why Single-Layer Perceptrons Fail on Non-Linear Problems (The XOR Gate)**

Consider the **XOR (Exclusive OR)** logical operation:
* Input: `[0, 0]` $\rightarrow$ Output: `0`
* Input: `[0, 1]` $\rightarrow$ Output: `1`
* Input: `[1, 0]` $\rightarrow$ Output: `1`
* Input: `[1, 1]` $\rightarrow$ Output: `0`

**The Limitation:**
If you plot these points on a 2D graph, you cannot draw a **single straight line** that separates the outputs of `1` from the outputs of `0`. Because a Single-Layer Perceptron can only create a linear decision boundary (a straight line), it fails completely. To solve this, a **Multi-Layer Perceptron (MLP)** with hidden layers is required to bend or combine decision boundaries.
