## 16. Loss Functions, Weight Updation, and Optimizers

### **1. Loss Functions**
A loss function measures how well a neural network model performs by calculating the error between the predicted value ($\hat{y}$) and the actual target ($y$).

* **Regression Loss Functions:**
  1. **MSE (Mean Squared Error):** Computes the average squared difference between estimated values and the actual value. Formula: $(y - \hat{y})^2$
  2. **MAE (Mean Absolute Error):** Computes the average absolute difference.
  3. **Huber Loss:** Combines the best properties of MSE and MAE, making it less sensitive to outliers.

* **Classification Loss Functions:**
  1. **Binary Cross-Entropy:** Used for binary classification tasks (e.g., Pass/Fail, 0 or 1).
  2. **Categorical Cross-Entropy:** Used for multi-class classification problems.


### **2. Weight Updation Formula (Gradient Descent)**
During backpropagation, weights are updated to minimize the loss using the gradient descent rule:

$$W_{new} = W_{old} - \eta \left( \frac{\partial L}{\partial W_{old}} \right)$$

* **$W_{new}$ & $W_{old}$:** The updated and previous weights.
* **$\eta$ (Eta / Learning Rate):** A hyperparameter that controls the step size taken during optimization.
* **$\frac{\partial L}{\partial W_{old}}$:** The derivative (slope/gradient) of the loss function with respect to the weight.


### **3. The Role of Optimizers & Learning Rate ($\eta$)**
**Optimizers** are algorithms used to change the attributes of the neural network (such as weights and learning rate) to reduce the overall loss and reach the **Global Minima** (the point where loss is lowest).

#### **Impact of Learning Rate Size:**
* **Small Learning Rate (e.g., $0.001$):** The model takes tiny steps down the loss curve, resulting in very safe but slow convergence.
* **Large Learning Rate:** The model takes huge leaps across the loss curve. As shown in the lecture graph, it can bounce back and forth wildly across the curve ("overshooting") and **may never converge** to the global minimum.

## 17. Example: Numerical Weight Updation (Gradient Descent)

To understand how the network learns, let’s manually perform one iteration of weight update for a single weight ($W_7$) based on the formula:
$$W_{new} = W_{old} - \eta \left( \frac{\partial L}{\partial W_{old}} \right)$$

### **Scenario Setup**
* **Initial Weight ($W_{old}$):** $0.5$
* **Learning Rate ($\eta$):** $0.1$
* **Calculated Slope (Gradient $\frac{\partial L}{\partial W}$):** $2.0$ (This represents the steepness of the error surface at the current point).


### **Calculation**
1. **Identify the values:**
   * $W_{old} = 0.5$
   * $\eta = 0.1$
   * $\text{Gradient} = 2.0$

2. **Apply the Formula:**
   $$W_{new} = 0.5 - (0.1 \times 2.0)$$
   $$W_{new} = 0.5 - 0.2$$
   $$W_{new} = 0.3$$

### **Interpretation**
* **The Direction:** Because the slope (gradient) was positive ($+2.0$), the formula subtracted from the weight. This moves the weight in the opposite direction of the slope, effectively heading toward the **Global Minima** (lower loss).
* **The Learning Rate:** If our learning rate ($\eta$) had been too large (e.g., $1.0$), the update would have been $W_{new} = 0.5 - (1.0 \times 2.0) = -1.5$, which is a massive jump that might have caused us to overshoot the target entirely, just as we discussed in the "large learning rate" scenario.
