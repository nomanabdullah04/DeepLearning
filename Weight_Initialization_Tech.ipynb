## 44. Weight Initialization Techniques & Key Principles

Proper weight initialization is crucial when building neural networks to prevent issues like vanishing or exploding gradients and to ensure effective learning.

### **1. Key Points / Core Principles of Weight Initialization**
As outlined in the lecture slide, effective weight initialization must satisfy three primary conditions:
* **Weights Should Be Small:** Initializing weights with excessively large values causes signals to blow up or saturate activation functions early in training.
* **Weights Should Not Be the Same:** If all weights in a layer start at the exact same value (e.g., all zeros or all ones), every neuron in that layer will compute the exact same output and receive the exact same gradient update during backpropagation, failing to break symmetry.
* **Weights Should Have Good Variance:** Maintaining a proper variance across weights ensures that signals can flow forward and backward through deep layers without vanishing or exploding.


### **2. Common Weight Initialization Techniques**
* **Uniform Distribution:** Initializes weights by drawing values randomly from a uniform probability distribution within a specific range.
* **Xavier / Glorot Initialization:** Specifically designed for activation functions like **Sigmoid and Tanh**. It scales the weight variance based on the number of input and output connections to keep the scale of gradients roughly constant across layers.
* **Kaiming He Initialization:** Designed specifically for **ReLU and its variants** (like Leaky ReLU or ELU). It accounts for the fact that ReLU zeroes out half of the inputs, scaling the variance to prevent vanishing or exploding gradients in deep networks.

## 45. Weight Initialization: Uniform Distribution

As shown in the lecture slide, the **Uniform Distribution** method is a classic approach for initializing neural network weights ($W_{ij}$) randomly within a controlled range.

### **1. Mathematical Formula & Bounds**
Instead of drawing random values from an arbitrarily wide range (which can cause exploding or vanishing gradients), weights are initialized uniformly between specific negative and positive boundaries based on the number of inputs:

$$W_{ij} \sim \text{Uniform} \left( -\frac{1}{\sqrt{n_{input}}}, \frac{1}{\sqrt{n_{input}}} \right)$$

* **Where:**
  * $W_{ij}$ represents the weight connection between nodes.
  * $n_{input}$ (or $\sqrt{input}$ as noted in the slide) is the number of input connections (fan-in) coming into the neuron.
  * The distribution ensures that weights are sampled evenly between the lower bound $-\frac{1}{\sqrt{n_{input}}}$ and the upper bound $+\frac{1}{\sqrt{n_{input}}}$.


### **2. Specific Example Scale**
* For instance, if the scaling factor resolves or is chosen such that $\sqrt{input} \approx \sqrt{3}$, the bounds simplify to:
  $$\left[ -\frac{1}{\sqrt{3}}, \frac{1}{\sqrt{3}} \right]$$

### **3. Why Use It?**
* **Keeps Initial Values Small:** By dividing by $\sqrt{n_{input}}$, the initial weights automatically scale down as the number of inputs increases, preventing activations from blowing up in the first layer.
* **Breaks Symmetry:** Random uniform sampling ensures that no two neurons start with the exact same weight values, allowing them to learn different features during backpropagation.

## 46. Xavier / Glorot Initialization: Normal and Uniform

**Xavier (or Glorot) Initialization** is designed specifically for neural networks using symmetric activation functions like **Sigmoid** or **Tanh**. It scales the initial weights based on the number of input connections (fan-in) and output connections (fan-out) to keep the variance of activations consistent across layers.

### **1. Xavier Normal Initialization**
Weights are drawn from a normal (Gaussian) distribution with a mean of $0$ and a calculated standard deviation ($\sigma$).

* **Formula:**
  $$W_{ij} \sim N(0, \sigma)$$
* **Standard Deviation ($\sigma$):**
  $$\sigma = \sqrt{\frac{2}{\text{input} + \text{output}}}$$
  *(Where $\text{input}$ is the number of incoming connections and $\text{output}$ is the number of outgoing connections).*


### **2. Xavier Uniform Initialization**
Weights are drawn uniformly from a symmetric range bounded by a calculated limit.

* **Formula:**
  $$W_{ij} \sim \text{Uniform} \left( -\frac{\sqrt{6}}{\sqrt{\text{input} + \text{output}}}, \frac{\sqrt{6}}{\sqrt{\text{input} + \text{output}}} \right)$$


### **Why Use Xavier Initialization?**
* **Stabilizes Variance:** Prevents the signal strength from exploding or vanishing as it passes through deep layers during the forward and backward passes.
* **Optimized for Sigmoid/Tanh:** Ensures that activations do not saturate completely at the extremes (0 or 1 for sigmoid, -1 or 1 for tanh) right at the start of training.

## 47. Kaiming He Initialization: Normal and Uniform

**Kaiming He Initialization** is designed specifically for neural networks using **ReLU and its variants** (like Leaky ReLU or ELU). Because ReLU activations zero out negative values, they reduce the variance of activations by half. He initialization accounts for this by scaling the weight variance based only on the number of input connections (fan-in).

### **1. He Normal Initialization**
Weights are drawn from a normal (Gaussian) distribution with a mean of $0$ and a calculated standard deviation ($\sigma$).

* **Formula:**
  $$W_{ij} \sim N(0, \sigma)$$
* **Standard Deviation ($\sigma$):**
  $$\sigma = \sqrt{\frac{2}{\text{input}}}$$
  *(Where $\text{input}$ represents the number of incoming connections to the neuron).*


### **2. He Uniform Initialization**
Weights are sampled uniformly from a symmetric range scaled by the number of inputs.

* **Formula:**
  $$W_{ij} \sim \text{Uniform} \left( -\frac{\sqrt{6}}{\sqrt{\text{input}}}, \frac{\sqrt{6}}{\sqrt{\text{input}}} \right)$$

### **Why Use He Initialization?**
* **Prevents Vanishing/Exploding Gradients in ReLU:** By scaling the variance by $\frac{2}{\text{input}}$, it compensates for the dead/zeroed-out negative space of ReLU neurons, ensuring stable signal and gradient flow across deep hidden layers.
