## 18. Backpropagation & the Chain Rule of Derivation

### **1. The Core Concept of Backpropagation**
Once the loss is computed at the output layer, **Backpropagation** flows backward from the output layer to the input layer to update weights.

To find how much a specific early weight (like $W_4$ or $W_1$) contributed to the final loss, we use the **Chain Rule of Calculus**.


### **2. Applying the Chain Rule (Single Path)**
For a weight near the output layer ($W_{4_{old}}$), the weight update formula is:
$$W_{4_{new}} = W_{4_{old}} - \eta \left( \frac{\partial L}{\partial W_{4_{old}}} \right)$$

Using the chain rule, the gradient breaks down as:
$$\frac{\partial L}{\partial W_{4_{old}}} = \frac{\partial L}{\partial O_2} \times \frac{\partial O_2}{\partial W_{4_{old}}}$$

For deeper weights (such as $W_{1_{old}}$ connecting back to Hidden Layer 1), the chain rule extends across multiple layers:
$$\frac{\partial L}{\partial W_{1_{old}}} = \frac{\partial L}{\partial O_2} \times \frac{\partial O_2}{\partial O_1} \times \frac{\partial O_1}{\partial W_{1_{old}}}$$


### **3. Handling Multiple Paths in Complex Networks (Assignments/Multi-Layer Architectures)**

image.png
In more complex network architectures (like the diamond-shaped multi-layered network shown in the final slides where a hidden layer splits into two separate paths—$O_{21}$ and $O_{22}$—before converging at $O_{31}$):

* **The Challenge:** The gradient must flow backward through **all possible paths** that share that weight.
* **The Solution:** According to calculus, when multiple paths exist, we compute the chain rule for each distinct path and **add them together ($+$)**:

$$\frac{\partial L}{\partial W_{1_{old}}} = \left( \frac{\partial L}{\partial O_{31}} \times \frac{\partial O_{31}}{\partial O_{21}} \times \frac{\partial O_{21}}{\partial O_{11}} \times \frac{\partial O_{11}}{\partial W_{1_{old}}} \right) + \left( \frac{\partial L}{\partial O_{31}} \times \frac{\partial O_{31}}{\partial O_{22}} \times \frac{\partial O_{22}}{\partial O_{12}} \times \frac{\partial O_{12}}{\partial W_{1_{old}}} \right)$$

This elegant application of the chain rule allows neural networks of any size or depth to calculate gradients accurately and update their weights during training!
